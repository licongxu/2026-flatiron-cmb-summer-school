## Introduction: B-modes and Primordial Gravitational Waves

From the [CMB Polarization notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_CMB_Polarization.ipynb), we have learnt that the polarization field of the CMB can be uniquely decomposed into two components of opposite parity: **E-modes** (gradient-like, curl-free patterns) and **B-modes** (curl-like, divergence-free patterns). This decomposition is powerful because of a simple physical fact: **scalar (density) perturbations — the ones that seed all cosmic structure — produce no B-modes at linear order**. Any primordial degree-scale $B$-mode is therefore a window onto other physics, beyond the scalar density perturbations that dominate the temperature and $E$-mode skies. 

The most exciting potential source of primordial B-modes is a background of **primordial gravitational waves (PGWs)**: tensor perturbations of the metric that essentially all models of **cosmic inflation** predict, generated by quantum fluctuations of spacetime itself during a rapid exponential expansion at the beginning of the Universe. As these gravitational waves stream through the plasma at recombination, they imprint a characteristic B-mode polarization pattern peaking at degree angular scales.

The amplitude of this signal is parameterized by **the tensor-to-scalar ratio $r$**: the ratio of the primordial power in tensor perturbations to that in scalar perturbations at a pivot scale, conventionally $k = 0.05\,\mathrm{Mpc}^{-1}$. This number is powerful because it turns a *pattern on the sky into a measurement of the physics of the very early Universe*. In simple inflationary models, $r$ is directly tied to the expansion rate and energy scale during inflation,
$$
V^{1/4} \approx 1.04 \times 10^{16}\,\mathrm{GeV}\,\left(\frac{r}{0.01}\right)^{1/4}.
$$
A **detection of primordial $B$-modes** ($r > 0$) would be evidence that inflation happened, but it would also mean that we had seen the faint imprint of gravitational waves produced when the Universe was almost unimaginably young. In that sense, the CMB would be acting like a fossil record of spacetime itself. In the inflationary picture, those waves began as quantum fluctuations of spacetime itself, later stretched across the sky by cosmic expansion. The size of the signal, described by $r$, would tell us how energetic and rapid that early expansion was — probing physics at energies far beyond anything we can recreate in a laboratory, and providing one of the few observational windows onto the quantum origin of spacetime fluctuations. It would also help distinguish between different possible versions of inflation, showing which kinds of early-Universe physics are still viable and which are ruled out. 


![](https://images.newscientist.com/wp-content/uploads/2019/02/mg32180201.jpg)

### Why is this measurement so hard? 

Current data already require this signal to be very small. The most stringent upper limits, from the [BICEP/Keck + Planck + WMAP](https://arxiv.org/pdf/2110.00483) gives $r_{0.05}<0.035$ at 95% confidence. Upcoming experiments, such as the Simons Observatory and LiteBIRD aim to push sensitivity to the $\sigma(r)\sim 0.001-0.003$ level, where either a detection would open a direct observational window onto ultra-high-energy physics, or a non-detection would rule out large families of otherwise plausible early-Universe models.

If the signal exists at the currently allowed level, it is buried under everything else:

1. **CMB temperature and E-mode polarization** – scalar density perturbations produce the dominant temperature anisotropies and E-mode polarization, but no B-modes at linear order. This is what makes B-modes such a clean target in principle — but in practice it also means that any leakage from T or E into B, due to masking, filtering, beam effects or calibration errors, has to be controlled extremely carefully.
2. **Gravitational lensing** of E-modes by large-scale structure converts a small fraction of E into B, producing a guaranteed "lensing B-mode" signal that dominates over the primordial one at $\ell \gtrsim 100$ (see the [Lensing Simulation](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Simulation.ipynb) and [Lensing Reconstruction](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Reconstruction.ipynb) notebooks).
3. **Galactic foregrounds** — polarized thermal dust and synchrotron emission from our own Galaxy — are brighter than the primordial signal at *every* frequency and over most of the sky.
4. **Instrumental noise and systematics** must be controlled at the nanokelvin level (see the [Modeling Instrumental Noise and Beams](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_modeling_noise_and_beams.ipynb) and [Real Experiment Beams](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_Beam_Systematics.ipynb) notebooks).

## Section 1: The CMB B-mode Power Spectrum

At the level allowed by the data, any primordial B-mode signal generated by PGWs, parameterized by the tensor-to-scalar ratio $r$, is mostly dominated by a lensing B-mode signal, due to gravitational lensing of E-modes into B-modes, parameterized by the lensing amplitude $A_\mathrm{lens}$ (=1 in standard $\Lambda$CDM) (see the [Lensing Simulation notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_Lensing_Simulation.ipynb) for where this signal comes from).

The total CMB B-mode angular power spectrum is:
$$D_\ell^{BB} = r \cdot D_\ell^{BB,\mathrm{tens}} + A_\mathrm{lens} \cdot D_\ell^{BB,\mathrm{lens}}$$

where $D_\ell = \ell(\ell+1)C_\ell / (2\pi)$ $[\mu K^2]$, and:
- $D_\ell^{BB,\mathrm{tens}}$ is the primordial tensor contribution for $r=1$
- $D_\ell^{BB,\mathrm{lens}}$ is the lensing B-mode spectrum

````python
# ============================================================
# 1.1  Compute CMB power spectra with CAMB
# ============================================================

lmax_camb = 1000

# Standard LCDM cosmology
cosmo_params = dict(H0=67.5, ombh2=0.022, omch2=0.122, mnu=0.06, omk=0, tau=0.06)

# --- Lensed scalar BB (r=0, gives lensing B-modes) ---
# This part is the same as what is done in CMB_School_Simulating_the_CMB.ipynb
pars0 = camb.CAMBparams()
pars0.set_cosmology(**cosmo_params)
pars0.WantTensors = False
pars0.InitPower.set_params(As=2e-9, ns=0.965, r=0)
pars0.set_for_lmax(lmax_camb, lens_potential_accuracy=1)
results0 = camb.get_results(pars0)
cls_lens = results0.get_lensed_scalar_cls(lmax=lmax_camb, CMB_unit='muK')
# Shape: (lmax_camb+1, 4), columns TT, EE, BB, TE; index = ell

# --- Tensor BB for r=1 ---
# We add the tensor component
pars1 = camb.CAMBparams()
pars1.set_cosmology(**cosmo_params)
pars1.WantTensors = True
pars1.InitPower.set_params(As=2e-9, ns=0.965, r=1, nt=0)
pars1.set_for_lmax(lmax_camb, lens_potential_accuracy=0)
results1 = camb.get_results(pars1)
cls_tens = results1.get_tensor_cls(lmax=lmax_camb, CMB_unit='muK')
# Tensor D_ell for r=1

# Store D_ell arrays indexed by ell (row index = ell)
ell_camb   = np.arange(lmax_camb + 1)   # 0..lmax_camb
Dl_lens_BB = cls_lens[:, 2]             # D_ell^BB from lensing
Dl_tens_BB = cls_tens[:, 2]             # D_ell^BB from tensors (r=1)

# Interpolation functions for arbitrary ell
f_Dl_lens = interp1d(ell_camb, Dl_lens_BB, bounds_error=False, fill_value=0.)
f_Dl_tens = interp1d(ell_camb, Dl_tens_BB, bounds_error=False, fill_value=0.)

def cmb_Dl_BB(ell, r=0.01, A_lens=1.0):
    """Total CMB D_ell^BB [muK^2] for given r and lensing amplitude."""
    return r * f_Dl_tens(ell) + A_lens * f_Dl_lens(ell)

def cmb_Cl_BB(ell, r=0.01, A_lens=1.0):
    """Total CMB C_ell^BB [muK^2] = D_ell * 2*pi / (ell*(ell+1))."""
    Dl = cmb_Dl_BB(ell, r, A_lens)
    return Dl * 2 * np.pi / (ell * (ell + 1))

print(f'CAMB spectra computed up to ell={lmax_camb}')
print(f'Peak lensing D_ell at ell ~ {ell_camb[np.argmax(Dl_lens_BB)]}')
````

````python
# ============================================================
# 1.2  Plot the CMB B-mode power spectrum
# ============================================================

ell_plot = np.arange(2, 400)

fig, ax = plt.subplots(figsize=(7, 5))

ax.plot(ell_plot, f_Dl_lens(ell_plot), 'k--', lw=2.5, label=r'Lensing BB ($A_{\rm lens}=1$)')
for r_val, color in [(0.1, 'C0'), (0.01, 'C1'), (0.001, 'C3')]:
    ax.plot(ell_plot,
            r_val * f_Dl_tens(ell_plot),
            color=color, label=f'Primordial tensors, $r={r_val}$')

ax.set_yscale('log')
ax.set_xscale('log')
ax.set_xlabel(r'Multipole $\ell$')
ax.set_ylabel(r'$D_\ell^{BB}$ [$\mu{\rm K}^2$]')
ax.set_title('CMB B-mode Power Spectrum')
ax.axvspan(30, 300, alpha=0.07, color='grey', label='Ground-based analysis range')
ax.legend(loc='upper left')
ax.set_xlim(2, 400)
ax.set_ylim(1e-6, 0.2)
plt.tight_layout()
plt.show()
````

The primordial $B$-mode signal has a distinctive shape: it is expected to **peak on degree angular scales** – the "recombination bump" at $\ell \approx 80$, with another peak on the largest scales, $\ell \lesssim 10$, from reionization. But we do not know how large the signal should be. The parameter $r$ is currently constrained only by upper limits, and different inflationary models predict different amplitudes depending on different possible mechanisms for the earliest expansion of the Universe.

The primordial signal is below the lensing $B$-mode signal over much of the range $\ell \gtrsim 80$ for $r=0.01$. Both the primordial signal and lensing signal are **subdominant compared to galactic foreground emission** at most frequencies and angular scales. We need to measure the sky at multiple frequencies to separate them.

## Section 2: Galactic Foregrounds

Two galactic emission processes dominate the B-mode signal over most of the sky and frequency range. The frequency dependence of foreground emission is given by its Spectral Energy Distribution (SED), which tells us how the brightness or polarised intensity of a non-CMB component changes with observing frequency.

### 2a. Thermal Dust Emission
Magnetically aligned dust grains emit thermally polarized radiation, well described by a **modified blackbody** (MBB) spectral energy distribution:
$$f_d(\nu) \propto \nu^{\beta_d + 1} \cdot \frac{B_\nu(T_d)}{B_{\nu_0}(T_d)},$$
where $B_\nu(T)$ is the Planck function, $\beta_d \approx 1.54$, $T_d \approx 20$ K, and $\nu_0 = 353$ GHz (the Planck reference frequency). This SED rises steeply with frequency, making dust dominant above ~100 GHz.

### 2b. Synchrotron Emission
Relativistic electrons spiraling in the galactic magnetic field emit a **power-law** SED in Rayleigh-Jeans temperature:
$$f_s(\nu) \propto \left(\frac{\nu}{\nu_0}\right)^{\beta_s},$$
with $\beta_s \approx -3.0$ and $\nu_0 = 23$ GHz. Synchrotron dominates at low frequencies (<70 GHz).

#### Unit Conventions
We work throughout in $K_\mathrm{CMB}$ units: the signal is expressed as a fluctuation in CMB thermodynamic temperature. The conversion from Rayleigh-Jeans ($K_\mathrm{RJ}$) to $K_\mathrm{CMB}$ introduces a frequency-dependent factor:
$$g(\nu) = \frac{dT_\mathrm{CMB}}{dT_\mathrm{RJ}} = \left(\frac{h\nu}{k_\mathrm{B} T_\mathrm{CMB}}\right)^2 \frac{e^x}{(e^x - 1)^2}, \quad x = \frac{h\nu}{k_\mathrm{B} T_\mathrm{CMB}}.$$
The **CMB SED in $K_\mathrm{CMB}$ units is frequency-independent ($=1$), which is the key property we exploit for component separation**.

````python
# ============================================================
# Global choice: frequency bands for the toy likelihood
# ============================================================

# Edit this list to change the frequency coverage of the toy experiment.
# Examples:
#   np.array([93., 145.])              # only two CMB-like bands
#   np.array([93., 145., 220.])        # baseline teaching setup
#   np.array([93., 145., 220., 280.])  # add another dust-monitoring band
freqs = np.array([50., 93., 145., 220.])   # GHz


def make_band_names(freqs):
    """Human-readable labels for frequency bands."""
    return [f"{nu:.0f} GHz" for nu in np.asarray(freqs)]


def band_noise_uK_arcmin(freqs):
    """
    Toy polarization white-noise level for each band [uK arcmin].

    This is the standard way to quote map depth (see the Modeling
    Instrumental Noise and Beams notebook). The 90/150 GHz "CMB channels"
    are the deepest; higher-frequency dust-monitoring bands are noisier,
    but they still help because they give strong leverage on dust.
    Real noise ladders are steeper (e.g. Planck 353 GHz: ~500 uK arcmin);
    the full B-modes notebook uses realistic values.
    """
    freqs = np.asarray(freqs, dtype=float)
    levels = np.full(freqs.shape, 6.0)   # deep CMB channels
    levels[freqs < 70.]   = 12.0         # low-frequency (synchrotron) bands
    levels[freqs >= 200.] = 15.0         # dust-monitoring bands
    levels[freqs >= 270.] = 25.0
    levels[freqs >= 330.] = 40.0
    return levels


band_names = make_band_names(freqs)
n_freq = len(freqs)
band_noise_level = band_noise_uK_arcmin(freqs)

print(f"Using {n_freq} frequency channel(s): {band_names}")
print("Toy white-noise levels [uK arcmin]:", band_noise_level)
````

````python
# ============================================================
# 2.1  Foreground SED functions (in K_CMB units)
# ============================================================

def g_nu(nu_GHz):
    """K_RJ -> K_CMB unit conversion factor: g(nu) = (x^2 e^x)/(e^x-1)^2."""
    x  = h_P * nu_GHz * 1e9 / (k_B * T_CMB)
    ex = np.exp(x)
    return (x**2 * ex) / (ex - 1)**2

def dust_sed(nu_GHz, nu0_GHz=353., beta_d=1.54, T_d=20.):
    """
    Thermal dust SED in K_CMB, normalized to 1 at nu0.
    Modified blackbody: f(nu) ~ nu^(beta+1) * B_nu(T_d)  in K_RJ.
    Then converted to K_CMB via g(nu0)/g(nu).
    """
    x  = h_P * nu_GHz * 1e9 / (k_B * T_d)
    x0 = h_P * nu0_GHz * 1e9 / (k_B * T_d)
    sed_rj = (nu_GHz / nu0_GHz)**(beta_d + 1) * np.expm1(x0) / np.expm1(x)
    return sed_rj * g_nu(nu0_GHz) / g_nu(nu_GHz)

def synch_sed(nu_GHz, nu0_GHz=23., beta_s=-3.0):
    """
    Synchrotron SED in K_CMB, normalized to 1 at nu0.
    Power law in K_RJ: f(nu) ~ (nu/nu0)^beta_s.
    Then converted to K_CMB via g(nu0)/g(nu).
    """
    sed_rj = (nu_GHz / nu0_GHz)**beta_s
    return sed_rj * g_nu(nu0_GHz) / g_nu(nu_GHz)

# --- Verify: at reference frequencies, SEDs = 1 ---
print(f'dust_sed(353 GHz)  = {dust_sed(353.):.4f}  (expect 1.0)')
print(f'synch_sed(23 GHz)  = {synch_sed(23.):.4f}   (expect 1.0)')
print(f'CMB SED at all nu  = 1.0  (frequency-independent in K_CMB)')
````

````python
# ============================================================
# 2.2  Plot the dust SED in K_CMB units
# ============================================================

# Make the plotting range automatically include all chosen bands.
nu_min = min(20., freqs.min() - 10.)
nu_max = max(300., freqs.max() + 30.)
nu_arr = np.linspace(nu_min, nu_max, 400)

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))

# Left: dust vs CMB
ax = axes[0]
ax.semilogy(nu_arr, np.abs(dust_sed(nu_arr)), 'C1', lw=2,
            label=r'Thermal dust ($\beta_d=1.54,\,T_d=20$ K)')
ax.semilogy(nu_arr, np.abs(synch_sed(nu_arr)), 'C2', lw=2,
            label=r'Synchrotron ($\beta_s=-3$)')
ax.axhline(1.0, color='k', ls='--', label=r'CMB: flat in $K_{\rm CMB}$')
for nu in freqs:
    ax.axvline(nu, color='grey', ls=':', alpha=0.6)
    ax.text(nu, 2e-3, f'{nu:.0f}', rotation=90, va='bottom', ha='right', fontsize=8)
ax.set_xlabel(r'Frequency [GHz]')
ax.set_ylabel(r'SED amplitude')
ax.set_title('CMB is frequency-independent; foregrounds are not')
ax.legend(fontsize=9)
ax.set_xlim(nu_min, nu_max)

# Right: sensitivity to beta_d
ax2 = axes[1]
for beta_d_val, ls in [(1.3, '--'), (1.54, '-'), (1.8, ':')]:
    ax2.semilogy(nu_arr, np.abs(dust_sed(nu_arr, beta_d=beta_d_val)),
                 'C1', ls=ls, lw=2, label=f'$\\beta_d={beta_d_val}$')
for beta_s_val, ls in [(-2.7, '--'), (-3.0, '-'), (-3.3, ':')]:
    ax2.semilogy(nu_arr, np.abs(synch_sed(nu_arr, beta_s=beta_s_val)),
                 'C2', ls=ls, label=f'$\\beta_s={beta_s_val}$')
for nu in freqs:
    ax2.axvline(nu, color='grey', ls=':', alpha=0.6)
ax2.set_xlabel(r'Frequency [GHz]')
ax2.set_ylabel(r'Dust SED amplitude')
ax2.set_title(r'Changing $\beta$ changes the frequency scaling')
ax2.legend(fontsize=9)
ax2.set_xlim(nu_min, nu_max)

plt.tight_layout()
plt.show()
````

The left panel illustrates the key idea behind multi-frequency component separation. **In thermodynamic temperature units, $K_\mathrm{CMB}$, the CMB SED is frequency-independent (its SED is simply equal to 1), which is the key property we exploit for component separation**. Galactic foregrounds do not have this property. Synchrotron emission is brighter at low frequencies, while thermal dust emission becomes brighter at high frequencies. This difference in frequency dependence is what allows us, in principle, to separate the CMB from foreground emission.

If we knew the exact frequency dependence of every component, component separation would be straightforward. In practice, foregrounds are only approximately described by simple models. For example,spectral parameters may vary across the sky, as illustrated in the right panel. This means that a single global parameter is usually not enough to describe the full foreground signal. There are many ways to deal with this complexity, from simple parametric models to more flexible approaches that allow the foreground SEDs to vary across the sky. We will not explore those methods in detail here. 

> *From now on we use dust only as an example.*

Using a parametric power-spectrum model, we can write the dust contribution as a frequency-dependent SED factor multiplied by a spatially varying amplitude. For the amplitude, we use the model of [Azzoni+ 2021](https://arxiv.org/pdf/2011.11575) and parametrize it as a simple power-law
$
D_\ell^{\rm dust} = A_d\left(\frac{\ell}{\ell_0}\right)^{\alpha_d},
$
where $A_d$ controls the dust amplitude at the pivot scale $\ell_0$, and $\alpha_d$ controls how the dust power changes with angular scale.


````python
# ============================================================
# 2.3  Dust angular power spectrum (B-mode)
# ============================================================

# Power-law D_ell model: D_ell = A * (ell/ell0)^alpha
ell0 = 80.   # pivot scale

def dust_Dl_BB(ell, A_d=20., alpha_d=-0.4):
    """Dust BB power spectrum D_ell [muK^2] at the reference frequency 353 GHz."""
    return A_d * (ell / ell0)**alpha_d


def Dl_to_Cl(Dl, ell):
    """D_ell [muK^2] -> C_ell [muK^2]."""
    return Dl * 2 * np.pi / (ell * (ell + 1))


def Cl_to_Dl(Cl, ell):
    """C_ell [muK^2] -> D_ell [muK^2]."""
    return Cl * ell * (ell + 1) / (2 * np.pi)


def tensor_only_Dl_BB(ell, r=0.01):
    """
    Primordial tensor-only BB spectrum.

    The helper cmb_Dl_BB returns lensing + tensor, so we subtract the
    lensing-only curve to isolate the primordial contribution.
    """
    return cmb_Dl_BB(ell, r=r) - f_Dl_lens(ell)


# Decide ell range and assumed input r
ell_pl = np.arange(20, 400)
r_plot = 0.01

fig, ax = plt.subplots(figsize=(7, 5))
ax.loglog(ell_pl, cmb_Dl_BB(ell_pl, r=r_plot), 'k-',  lw=2.5,
          label=fr'CMB: lensing + tensor, $r={r_plot}$')
ax.loglog(ell_pl, f_Dl_lens(ell_pl), 'k--', lw=1.5,
          label=r'CMB: lensing only')
ax.loglog(ell_pl, tensor_only_Dl_BB(ell_pl, r=r_plot), 'k:', lw=2.0,
          label=fr'CMB: tensor only, $r={r_plot}$')

# Plot dust at every frequency currently selected in the setup cell.
for i, nu in enumerate(freqs):
    Dl_dust_nu = dust_Dl_BB(ell_pl) * dust_sed(nu)**2
    ax.loglog(ell_pl, Dl_dust_nu, label=f'Dust at {nu:.0f} GHz')

ax.set_xlabel(r'Multipole $\ell$')
ax.set_ylabel(r'$D_\ell^{BB}$ [$\mu{\rm K}^2$]')
ax.set_title('B-mode Power Spectra: CMB vs Dust')
ax.legend(fontsize=8)
ax.set_xlim(20, 400)
ax.set_ylim(1e-5, 1)
plt.tight_layout()
plt.show()
````

### Exercise 1 notes

1. **Varying $r$**: the primordial (tensor) BB spectrum scales linearly with $r$. For $r\lesssim 0.01$ it sits below the lensing BB over much of $\ell\gtrsim 80$, so measuring small $r$ needs either delensing, exquisite control of systematics, or both.

2. **Frequency bands**: dust BB $\propto f_d(\nu)^2$ rises steeply toward high frequency, while the CMB is flat in $K_{\rm CMB}$. Extra high-frequency channels are dust monitors: they pin down $A_d$ and $\beta_d$ even if they are noisier, which reduces the dust residual that would otherwise bias or broaden the $r$ posterior.

3. **Delensing ($A_{\rm lens}<1$)**: lowering the lensing floor reveals the recombination bump of the primordial signal. Perfect delensing ($A_{\rm lens}=0$) would *not* remove the need for multi-frequency observations — galactic dust is still brighter than $r\sim 0.01$ at most frequencies — but it does improve the raw sensitivity to $r$ once foregrounds are controlled.

## Section 3: Modelling the sky

For a pair of observing frequencies $\nu_i$ and $\nu_j$, we model the B-mode bandpower as a combination of CMB + Dust + noise:

$$
C_\ell^{\nu_i \times \nu_j}
=
C_\ell^{\rm CMB}(r)
+
f_d(\nu_i;\beta_d)\,f_d(\nu_j;\beta_d)\,C_\ell^{\rm dust}(A_d,\alpha_d) + N_\ell^{\nu_i \nu_j},
$$

where $f_d(\nu_i)$, is the dust SED at band $i$  This is defined by a set of parameters:
- $r$ controls the primordial B-mode amplitude;
- $A_d$ controls the dust amplitude;
- $\beta_d$ controls the dust frequency scaling;
- we fix the dust angular slope to $\alpha_d=-0.4$ and lensing amplitude to 1.

### The SED Mixing Matrix

The outer product $f(\nu_i) \cdot f(\nu_j)$ defines the **SED mixing matrix** for each component. For $N_\nu$ frequency bands, this is an $N_\nu \times N_\nu$ matrix that tells us how each component scales across frequency–frequency pairs. 

By fitting a parametric model to the multi-frequency data vector $\{C_\ell^{\nu_i \times \nu_j}\}$, we can simultaneously constrain $r$ and the foreground parameters.

This *parametric, power-spectrum-level* approach is complementary to the *blind, map-level* Internal Linear Combination (ILC) technique used in the [Y-map notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_ymap.ipynb): the ILC combines maps with weights that minimize the variance of the residual, without assuming a foreground model, while here we fit an explicit physical model for every component directly at the power-spectrum level.

````python
# ============================================================
# 3.1  Frequency bands and SED vectors
# ============================================================

# Refresh these derived quantities in case you edited freqs in the setup cell.
freqs = np.asarray(freqs, dtype=float)
band_names = make_band_names(freqs)
n_freq = len(freqs)
band_noise_level = band_noise_uK_arcmin(freqs)

# Fiducial dust spectral index
beta_d_fid = 1.54

# SED vectors at each band
f_cmb  = np.ones(n_freq)                    # CMB: frequency-independent
f_dust = dust_sed(freqs, beta_d=beta_d_fid) # dust: frequency-dependent

print(f'Number of frequency channels: {n_freq}')
print(f'Number of unique auto/cross spectra: {n_freq * (n_freq + 1) // 2}')
print('\nSED values at each band:')
print(f'{"Band":>10}  {"CMB":>8}  {"Dust":>10}  {"noise [uK arcmin]":>18}')
for i, nu in enumerate(freqs):
    print(f'{nu:>8.0f} GHz  {f_cmb[i]:>8.4f}  {f_dust[i]:>10.4f}  {band_noise_level[i]:>18.1f}')
````

## Section 4: Make a simple toy dataset

Here we build an intentionally simple toy dataset, to see how the likelihood recover a known value of $r$ while marginalizing over dust.

1. compute the true multi-frequency model;
2. build (simplified) diagonal error bars for each $D_\ell$ bandpower from a per-band white-noise level;
3. add one controlled noise realization;

````python
# ============================================================
# 4.1  Binning and dust-only signal model
# ============================================================

ell_min, ell_max = 5, 280
n_ell_bins = 10
ell_edges = np.linspace(ell_min, ell_max, n_ell_bins + 1)
ell_bins  = 0.5 * (ell_edges[:-1] + ell_edges[1:])
delta_ell = ell_edges[1] - ell_edges[0]   # bin width, used in the mode count

# Unique auto/cross frequency pairs, built automatically for any n_freq.
# For example, 3 bands -> 6 spectra; 4 bands -> 10 spectra.
i_triu, j_triu = np.triu_indices(n_freq)
n_cross = len(i_triu)

fit_param_names = ["r", "A_d", "beta_d"]

print(f"ell-bin centers: {ell_bins.astype(int)}")
print(f"frequency pairs: {n_cross}")
for icr, (i, j) in enumerate(zip(i_triu, j_triu)):
    print(f"  {icr:2d}: {band_names[i]} x {band_names[j]}")
print(f"data-vector length: {n_ell_bins * n_cross}")


def compute_Dl_model(r, A_d, beta_d, alpha_d=-0.4, A_lens=1.0, freqs_use=None):
    """
    Dust-only multi-frequency BB model in D_ell units.

    Parameters varied in the likelihood:
        r       : tensor-to-scalar ratio
        A_d     : dust BB amplitude at ell=80 and 353 GHz
        beta_d  : dust frequency spectral index

    Fixed in this teaching example:
        alpha_d : dust angular slope
        A_lens  : lensing B-mode amplitude
    """
    if freqs_use is None:
        freqs_use = freqs

    freqs_use = np.asarray(freqs_use, dtype=float)
    i_use, j_use = np.triu_indices(len(freqs_use))

    Dl_cmb = cmb_Dl_BB(ell_bins, r=r, A_lens=A_lens)
    Dl_dust_ref = dust_Dl_BB(ell_bins, A_d=A_d, alpha_d=alpha_d)
    f_d = dust_sed(freqs_use, beta_d=beta_d)

    model = np.zeros((n_ell_bins, len(i_use)))
    for icr, (i, j) in enumerate(zip(i_use, j_use)):
        model[:, icr] = Dl_cmb + f_d[i] * f_d[j] * Dl_dust_ref

    return model
````

The model has only three free parameters. The lensing amplitude and the dust angular slope are fixed, so the exercise focuses on the main idea: recover the frequency-independent primordial CMB contribution while marginalizing over the amplitude and frequency scaling of dust. Changing the number of frequency channels changes the length of the data vector, but not the parameter list.

#### Let's construct toy error bars

Each band is assigned a polarization white-noise level $\sigma_{\rm pol}$ in µK·arcmin — the standard way to quote map depth (see the [Modeling Instrumental Noise and Beams notebook](https://colab.research.google.com/github/flatironinstitute/2026-flatiron-cmb-summer-school/blob/main/notebooks/CMB_School_modeling_noise_and_beams.ipynb)). For simplicity, we assume no large scale $1/\ell$ noise for this toy example. White noise has a flat power spectrum,

$$
N_\ell = \left(\sigma_{\rm pol}\cdot \frac{\pi}{180\cdot 60}\right)^{2} \;[\mu K^2\,{\rm sr}],
$$
Each bin of width $\Delta\ell$ averages $N_{\rm modes} = (2\ell+1)\,\Delta\ell\, f_{\rm sky}$ modes, and the noise part of the error on the $\nu_i\times\nu_j$ bandpower is the noise-dominated limit of the Knox formula, roughly:

$$
\sigma_b^{\rm noise} = \sqrt{ \frac{(1+\delta_{ij})\, N^D_i\, N^D_j }{ N_{\rm modes} } } .
$$

The geometric mean $\sqrt{N^D_i N^D_j}$ is how the noise of two maps enters a cross-spectrum: a noisy dust-monitoring band crossed with a deep CMB band is penalized only by the *square root* of its noise — one reason cross-spectra are so useful. The $(1+\delta_{ij})$ doubles the variance of auto-spectra, where the same noise map appears twice. To this we add in quadrature a fractional term, standing in for the sample-variance part of the Knox formula: brighter spectra are measured with larger absolute errors. For simplicity, we neglect beams, signal–noise cross terms, and bin–bin correlations.

````python
# ============================================================
# 4.2  Generate a controlled synthetic dataset
# ============================================================

# This is a deliberately detectable teaching case.
# A smaller value such as r=0.003 would mostly give an upper limit in this toy setup.
params_true = dict(r=0.02, A_d=20.0, beta_d=1.54)
alpha_d_fixed = -0.4

model_true = compute_Dl_model(**params_true, alpha_d=alpha_d_fixed)

# Sky fraction observed (SO-SAT-like); sets the mode count per bin.
f_sky = 0.1
# Fractional error term: stand-in for the sample-variance part of the errors.
fractional_noise = 0.04


def noise_Dl_white(freqs_use=None):
    """
    White-noise D_ell = ell(ell+1)/(2 pi) * N_ell for each band.

    N_ell = (sigma_pol converted from uK arcmin to uK rad)^2 is flat in ell,
    so the noise rises as ell^2 in D_ell units.
    Returns an array of shape (n_ell_bins, n_bands).
    """
    if freqs_use is None:
        freqs_use = freqs
    levels = band_noise_uK_arcmin(freqs_use)
    N_ell = (levels * np.pi / (180. * 60.))**2   # uK^2 sr
    return ell_bins[:, None] * (ell_bins[:, None] + 1) / (2 * np.pi) * N_ell[None, :]


def make_sigma_Dl(model, freqs_use=None):
    """
    Diagonal error bars built from the white-noise levels (see text above):

        sigma_b^2 = (1 + delta_ij) * N^D_i * N^D_j / N_modes  +  (4% of model)^2

    with N_modes = (2 ell + 1) * delta_ell * f_sky.
    """
    if freqs_use is None:
        freqs_use = freqs
    i_use, j_use = np.triu_indices(len(freqs_use))
    N_Dl = noise_Dl_white(freqs_use)
    N_modes = (2 * ell_bins + 1) * delta_ell * f_sky
    sigma = np.zeros_like(model)

    for icr, (i, j) in enumerate(zip(i_use, j_use)):
        auto_factor = 1.0 + (i == j)   # auto-spectra: the noise is counted twice
        white = np.sqrt(auto_factor * N_Dl[:, i] * N_Dl[:, j] / N_modes)
        fractional = fractional_noise * np.abs(model[:, icr])
        sigma[:, icr] = np.sqrt(white**2 + fractional**2)

    return sigma


sigma_Dl = make_sigma_Dl(model_true)

# Add a controlled random realization.
# The rescaling makes chi2/dof close to 1 at the true parameters, which is easier for teaching.
rng = np.random.default_rng(8)
z = rng.standard_normal(model_true.size)
n_dof_target = model_true.size - len(fit_param_names)
z *= np.sqrt(n_dof_target / np.sum(z**2))
noise_realization = z.reshape(model_true.shape) * sigma_Dl

data_Dl = model_true + noise_realization

print(f"Frequency channels: {band_names}")
print(f"White-noise levels [uK arcmin]: {band_noise_level}")
print(f"Number of data points: {data_Dl.size}")
print(f"Number of fitted parameters: {len(fit_param_names)}")
print(f"Expected degrees of freedom: {n_dof_target}")
print(f"Injected parameters: r={params_true['r']}, A_d={params_true['A_d']}, beta_d={params_true['beta_d']}")
````

````python
# ============================================================
# 4.3  Plot one spectrum from the toy dataset
# ============================================================
def get_cross_index(fi, fj, nfreq=None):
    """Return the upper-triangle index for frequency pair fi x fj."""
    if nfreq is None:
        nfreq = n_freq
    if fi > fj:
        fi, fj = fj, fi
    ii, jj = np.triu_indices(nfreq)
    for k, (a, b) in enumerate(zip(ii, jj)):
        if a == fi and b == fj:
            return k
    raise ValueError(f"Could not find cross-spectrum {fi} x {fj}")


def plot_toy_spectrum(fi=0, fj=1, model=None, title_suffix=""):
    """Plot one D_ell spectrum from the toy data."""
    if model is None:
        model = model_true

    if fi >= n_freq or fj >= n_freq:
        raise ValueError(f"Requested band index out of range. This setup has n_freq={n_freq}.")

    icr = get_cross_index(fi, fj)
    nu_i, nu_j = freqs[fi], freqs[fj]

    fig, ax = plt.subplots(figsize=(6, 4.5))
    ax.errorbar(
        ell_bins, data_Dl[:, icr], yerr=sigma_Dl[:, icr],
        fmt="o", color="k", ms=4, label="toy data"
    )
    ax.plot(ell_bins, model_true[:, icr], lw=2, label="true total model")

    Dl_cmb = cmb_Dl_BB(ell_bins, r=params_true["r"], A_lens=1.0)
    f_i = dust_sed(nu_i, beta_d=params_true["beta_d"])
    f_j = dust_sed(nu_j, beta_d=params_true["beta_d"])
    Dl_dust = f_i * f_j * dust_Dl_BB(ell_bins, params_true["A_d"], alpha_d_fixed)

    ax.plot(ell_bins, Dl_cmb, "--", label="CMB: lensing + primordial")
    ax.plot(ell_bins, Dl_dust, ":", label="dust")

    if title_suffix:
        ax.plot(ell_bins, model[:, icr], "--", lw=2, label=title_suffix)

    ax.set_title(f"{band_names[fi]} x {band_names[fj]}")
    ax.set_xlabel(r"$\ell$")
    ax.set_ylabel(r"$D_\ell^{BB}$ [$\mu K^2$]")
    ax.set_xlim(ell_min - 10, ell_max + 10)
    ax.legend(fontsize=8)
    plt.tight_layout()
    plt.show()

# Plot a representative cross-spectrum.
if n_freq >= 2:
    plot_toy_spectrum(fi=1, fj=1)
else:
    plot_toy_spectrum(fi=0, fj=0)
````

## Section 5: A simple likelihood for $r$ and dust

Because this is a teaching mock, we use a deliberately simple covariance model. The error bars are diagonal, which means that each plotted data point is treated as having its own independent uncertainty. In this case, comparing the model to the data reduces to a weighted sum of squared residuals.

For a given choice of parameters $(r,A_d,\beta_d)$, we compute
$$
\chi^2(r,A_d,\beta_d)
=
\sum_b \sum_{i\le j}
\left[
\frac{
D_{b,\mathrm{data}}^{\nu_i\times\nu_j}
-
D_{b,\mathrm{model}}^{\nu_i\times\nu_j}(r,A_d,\beta_d)
}
{\sigma_b^{\nu_i\times\nu_j}}
\right]^2.
$$
The sum runs over multipole bins $b$ and over all unique frequency cross-spectra $\nu_i\times\nu_j$. Each term measures the difference between the mock data and the model prediction in units of the corresponding error bar.

For independent Gaussian errors, the likelihood is directly related to $\chi^2$:
$$
\mathcal{L}(r,A_d,\beta_d)
\propto
\exp\left[
-\frac{1}{2}\chi^2(r,A_d,\beta_d)
\right].
$$
Therefore, models with smaller $\chi^2$ have larger likelihood. Minimizing $\chi^2$ is equivalent to finding the maximum-likelihood model.

We also need to specify priors for the parameters, $\pi(r,A_d,\beta_d)$. A prior is the information or assumptions we impose about the parameters before looking at the data. In this example we use simple flat, or “top-hat”, priors:
$$
0 \leq r \leq 0.06,
0 \leq A_d \leq 80,
1.0 \leq \beta_d \leq 2.2.
$$
Values inside these ranges are allowed equally; values outside are assigned zero probability. In the code this is implemented by returning 0 for allowed parameter values and $-\infty$ for disallowed values in log-probability space.
The posterior $\mathcal{P}$ is then $\propto$ likelihood × prior,
$$
\ln \mathcal{P}(r,A_d,\beta_d \mid \mathrm{data})
=
\ln \mathcal{L}(\mathrm{data}\mid r,A_d,\beta_d)
+
\ln \pi(r,A_d,\beta_d)
$$
Because the priors are flat, the posterior has the same shape as the likelihood inside the allowed parameter range, but is cut off at the prior boundaries.

Intuitively, a small $\chi^2$ means that the model passes close to the data compared with the size of the error bars. A large $\chi^2$ means that the model misses the data by more than the uncertainties allow.

One important caution is that a weak constraint on $r$ is not necessarily a sampler failure. If the likelihood is nearly flat as a function of $r$, then the posterior will be dominated by the prior: the data simply do not contain enough information to measure $r$. In this first teaching example, we choose a mock dataset where the likelihood has a visible preference for $r$, so that the connection between the data, the model, the likelihood and the posterior is easier to see.


````python
# ============================================================
# 5.1  Prior, likelihood and posterior
# ============================================================

def log_prior(theta):
    """Flat priors for the three fitted parameters."""
    r, A_d, beta_d = theta

    if not (0.0 <= r <= 0.06):
        return -np.inf
    if not (0.0 <= A_d <= 80.0):
        return -np.inf
    if not (1.0 <= beta_d <= 2.2):
        return -np.inf

    return 0.0


def log_likelihood(theta, data=data_Dl, sigma=sigma_Dl, freqs_use=freqs):
    """Diagonal Gaussian likelihood in D_ell bandpowers."""
    r, A_d, beta_d = theta
    model = compute_Dl_model(r=r, A_d=A_d, beta_d=beta_d,
                             alpha_d=alpha_d_fixed, freqs_use=freqs_use)
    residual = (data - model) / sigma
    chi2 = np.sum(residual**2)
    return -0.5 * chi2


def log_posterior(theta, data=data_Dl, sigma=sigma_Dl, freqs_use=freqs):
    lp = log_prior(theta)
    if not np.isfinite(lp):
        return -np.inf
    return lp + log_likelihood(theta, data=data, sigma=sigma, freqs_use=freqs_use)


theta_true = np.array([params_true["r"], params_true["A_d"], params_true["beta_d"]])
chi2_true = -2 * log_likelihood(theta_true)
n_dof = data_Dl.size - len(theta_true)

print(f"chi2 at true parameters: {chi2_true:.1f}")
print(f"degrees of freedom:       {n_dof}")
print(f"reduced chi2:             {chi2_true / n_dof:.2f}")
````

A reduced $\chi^2$ near 1 means the scatter of the data around the model is about what we expect from the error bars. In this toy case that happens by construction; in a real analysis it is a diagnostic of the data.


````python
# ============================================================
# 5.2  A quick 1D likelihood scan in r
# ============================================================

#r_grid = np.linspace(0.0, 0.04, 100)
r_grid = np.linspace(-1.0, 0.05, 1000)
chi2_grid = []

# Hold dust fixed only for this diagnostic plot.
# The MCMC below varies dust and r together.
for r_value in r_grid:
    theta = np.array([r_value, params_true["A_d"], params_true["beta_d"]])
    chi2_grid.append(-2 * log_likelihood(theta))

chi2_grid = np.array(chi2_grid)

fig, ax = plt.subplots(figsize=(5.5, 4))
ax.plot(r_grid, chi2_grid - chi2_grid.min(), lw=2)
ax.axvline(params_true["r"], color="k", ls="--", label="true r")
ax.axhline(1.0, color='grey', ls=':', label=r'$\Delta\chi^2=1$ (1$\sigma$)')
ax.axhline(4.0, color='grey', ls='--', label=r'$\Delta\chi^2=4$ (2$\sigma$)')
ax.set_xlabel(r"$r$")
ax.set_ylabel(r"$\Delta\chi^2$")
ax.set_title(r"Likelihood slice for $r$ (all other parameters fixed at truth)")
ax.set_xlim(-0.01, 0.1)
ax.set_ylim(-0.5, 10)
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()
````

The scan above is useful, but incomplete: it pretends we know the dust parameters exactly (if we did, recovering $r$ would be much easier).
This is a 1D slice: in reality we need to marginalize over all FG parameters, and the true posterior on $r$ will be broader due to foreground parameter degeneracies.

The MCMC below does the real teaching step: it lets $r$, $A_d$, and $\beta_d$ move together.

## Section 6: Parameter Estimation with Markov Chain Monte Carlo (MCMC)

#### Why not just use a grid?

What we want to report is the **marginalized posterior** of $r$: the 3-dimensional posterior $P(r, A_d, \beta_d \,|\, \mathbf{d})$ integrated over the dust parameters,

$$
p(r\,|\,\mathbf{d}) = \int P(r, A_d, \beta_d \,|\, \mathbf{d})\; dA_d\, d\beta_d .
$$

With only three parameters, a grid with 100 points per axis already costs $10^6$ likelihood evaluations. A realistic B-mode likelihood has 8 or more parameters (add $A_{\rm lens}$, synchrotron, spectral running, systematics...), and $100^8 = 10^{16}$ evaluations is **centuries of computing**. Beyond 3–4 dimensions, grids are hopeless.

#### The idea

MCMC replaces the grid with a **random walk that spends its time in proportion to the posterior**: where the posterior is twice as high, the chain leaves twice as many samples. Once you have such samples, everything becomes easy:

- **marginalizing** = simply *ignoring the other columns* of the chain;
- the histogram of the $r$ column **is** $p(r\,|\,\mathbf{d})$;
- medians and credible intervals are ordinary sample statistics.

#### The Metropolis–Hastings recipe

1. Start somewhere: $\theta_0 = (r, A_d, \beta_d)_0$.
2. **Propose** a random step: $\theta' = \theta_t + \Delta\theta$, with $\Delta\theta$ drawn from a Gaussian with fixed widths (the *proposal*, or *step size*).
3. Compute the posterior ratio $a = P(\theta'\,|\,\mathbf{d})\,/\,P(\theta_t\,|\,\mathbf{d})$ (in logs, in practice).
4. **Accept** with probability $\min(1, a)$: if accepted, $\theta_{t+1} = \theta'$; if rejected, $\theta_{t+1} = \theta_t$ — *repeat the current point in the chain* (essential, don't skip it!).
5. Go to 2.

Two features make this work:

- **Uphill moves are always accepted; downhill moves sometimes.** This is what lets the chain trace out the full *width* of the posterior instead of climbing to the peak and staying there — an optimizer finds the best fit, but MCMC measures the uncertainty.
- **Only posterior *ratios* appear**, so we never need the (unknown) overall normalization of the posterior.

Three practical rules of thumb, each of which we will *see* in the plots below:

- **Step size**: aim for roughly 20–40% of proposals accepted (too big → almost everything is rejected and the chain sits still; too small → everything is accepted but the chain crawls).
- **Burn-in**: throw away the beginning of the chain, which still remembers the arbitrary starting point.
- **Convergence**: run several chains from different starting points and check that they agree (the Gelman–Rubin statistic $\hat{R} \approx 1$).

In real data analyses, before starting the chain one should first find a good-fit point by minimizing the negative log-posterior. This maximizes the posterior probability, i.e. if a proposed parameter value lies lies outside the prior range, the posterior is $-\infty$, and the code returns a very large number to tell the optimizer to avoid that region. This makes the traces easier to read: the MCMC then spends its time exploring the posterior rather than slowly walking in from a bad starting point. 

<font color='red'>EXERCISE 3:</font> **Find a good-fit point before running the MCMC**. Write an objective function that an optimizer can minimize in order to find the maximum-a-posteriori values of $r$, $A_d$, and $\beta_d$. Remember that maximizing the log-posterior is equivalent to minimizing its negative, and that points outside the prior range should be rejected.
Try different initial guesses. Does the optimizer return the same values? Compare the result with the true input parameters, and explain why they may not agree exactly even for simulated data.

````python
# --------------------------------------------------------------
# EXERCISE 3: Find the maximum-a-posteriori (MAP) point
# --------------------------------------------------------------
# Maximize log_posterior  <=>  minimize -log_posterior.
# Outside the prior, log_posterior = -inf; return a huge value
# so the optimizer stays inside the allowed region.

def neg_log_posterior(theta):
    """Objective for scipy.optimize.minimize."""
    lp = log_posterior(theta)
    if not np.isfinite(lp):
        return 1e20
    return -lp


# Try several different starting points.
initial_guesses = [
    np.array([0.01, 15.0, 1.40]),   # low-ish r, low dust
    np.array([0.03, 30.0, 1.70]),   # high-ish r, high dust
    np.array([0.005, 10.0, 1.20]),  # near prior edge
    np.array([0.02, 20.0, 1.54]),   # near truth
]

map_results = []
for ig, x0 in enumerate(initial_guesses):
    res = minimize(neg_log_posterior, x0, method='Nelder-Mead')
    theta_hat = res.x
    map_results.append((x0, theta_hat, res.fun, res.success))
    print(f'Guess {ig}: start = {np.round(x0, 4)}')
    print(f'         MAP   = {np.round(theta_hat, 5)}  '
          f'(-logP = {res.fun:.3f}, success={res.success})')

# Use the best (lowest -logP) among the starts.
best = min(map_results, key=lambda t: t[2])
theta_map = np.array(best[1], dtype=float)

print('\nBest MAP among starts:')
for name, val, true in zip(fit_param_names, theta_map, theta_true):
    print(f'  {name:7s}:  MAP = {val:.5f}   true = {true:.5f}   '
          f'diff = {val - true:+.5f}')

print(
    '\nWhy MAP need not equal the true inputs even for simulated data:\n'
    '  - the toy data include one random noise realization;\n'
    '  - flat priors cut the posterior near boundaries;\n'
    '  - dust–r degeneracies shift the peak slightly along the ridge;\n'
    '  - the diagonal covariance is an approximation (not the full Knox cov).'
)

# Optional: confirm chi2 at MAP vs truth
print(f'\nchi2(MAP)  = {-2 * log_likelihood(theta_map):.2f}')
print(f'chi2(true) = {-2 * log_likelihood(theta_true):.2f}')
````

````python
# ============================================================
# 6.1  Metropolis-Hastings sampler
# ============================================================

def run_mcmc(log_probability, start, step_size, n_steps=20000, seed=0):
    """A compact Metropolis-Hastings sampler.

    The numbered comments match the recipe in the text above.
    """
    rng = np.random.default_rng(seed)

    current = np.array(start, dtype=float)      # (1) start somewhere
    logp_current = log_probability(current)

    chain = np.zeros((n_steps, len(current)))
    n_accept = 0

    for i in range(n_steps):
        # (2) propose a Gaussian random step
        proposal = current + step_size * rng.standard_normal(len(current))
        logp_proposal = log_probability(proposal)

        # (3)+(4) accept with probability min(1, posterior ratio).
        # In logs: accept if log(u) < log P(proposal) - log P(current).
        if np.log(rng.random()) < (logp_proposal - logp_current):
            current = proposal
            logp_current = logp_proposal
            n_accept += 1
        # if rejected: 'current' is unchanged and gets repeated below

        chain[i] = current                      # (5) record, then go to (2)

    return chain, n_accept / n_steps
````

````python
# ============================================================
# 6.2  Run a few Metropolis-Hastings chains
# ============================================================
n_chains = 4
n_steps = 18000
burn_in = 3000

# Proposal widths: a rough guess of each parameter's posterior width,
# times a factor ~0.4, so that EVERY parameter moves a healthy fraction of
# its own width at each proposal. This per-parameter matching is what keeps
# all three traces mixing well (Exercise 5 explores what happens otherwise).
# If you change the noise levels or the bands, update sigma_guess to match.
sigma_guess = np.array([0.005, 0.6, 0.02])   # rough widths of (r, A_d, beta_d)
step_size = 0.4 * sigma_guess

rng_start = np.random.default_rng(123)
start_scatter = np.array([0.002, 1.0, 0.04])

chains = []
acceptances = []

for ichain in range(n_chains):
    start = theta_map + start_scatter * rng_start.standard_normal(3)
    start[0] = np.clip(start[0], 0.001, 0.055)
    start[1] = np.clip(start[1], 1.0, 79.0)
    start[2] = np.clip(start[2], 1.02, 2.18)

    chain_i, acc_i = run_mcmc(
        log_posterior,
        start=start,
        step_size=step_size,
        n_steps=n_steps,
        seed=10 + ichain,
    )
    chains.append(chain_i)
    acceptances.append(acc_i)

chains = np.array(chains)  # shape: (n_chains, n_steps, n_parameters)
acceptances = np.array(acceptances)

# Combine all post-burn-in samples.
samples = chains[:, burn_in:, :].reshape(-1, len(fit_param_names))

print(f"Mean acceptance fraction: {acceptances.mean():.2f}")
print(f"Acceptance fractions: {np.round(acceptances, 2)}")
print(f"Samples after burn-in: {len(samples)}")
````

A healthy acceptance fraction is not a magic number, but values of order 0.2–0.6 are usually fine for this simple example. Beware, though: the overall acceptance fraction alone can hide problems. It is dominated by the well-tuned directions, so a single badly-scaled parameter can mix very poorly while the acceptance still looks healthy — this is why the proposal width is matched to each parameter separately above, and why we always check the per-parameter trace plots and $\hat{R}$ below. Before compressing the chains into numbers, let's dissect what the sampler actually did.

#### Watching the sampler work

The stored animation is used to "dissect" the algorithm. The grey contours show the posterior surface in the $(r,A_d)$ plane, evaluated at the best-fit value of $\beta_d$. This 2D slice is easy to draw here, but the same grid-based approach quickly becomes impractical in higher dimensions. This is exactly where MCMC becomes useful.

1. **the random walk** - The first part of the animation follows one chain, deliberately initialized away from the high-probability region. Accepted moves are shown as connected points, while rejected proposals are marked with red crosses. When a proposal is rejected, the chain stays at its current position, so the current point is repeated in the sample chain. The initial motion toward the posterior peak is the transient phase removed by burn-in; after that, the chain has effectively forgotten its starting point.
2. **the sample "cloud"** - After burn-in, samples from all chains form a cloud whose density follows the posterior. Regions with higher posterior probability contain more samples. In this way, the chains trace the posterior distribution without requiring a full parameter-space grid (no grid was ever built, yet the samples map out the same contours — and they would keep doing so with 8 or 20 parameters, where drawing the grey contours is impossible).

````python
# --------------------------------------------------------------
# EXERCISE 6: Design-your-experiment mini study
# --------------------------------------------------------------
# Compare three frequency configurations with a *short* MCMC each,
# reusing the same true sky (r, A_d, beta_d) and noise model.
# This answers: how does the r posterior change with frequency coverage?

def make_toy_for_freqs(freqs_use, params=params_true, seed=8):
    """Build (data, sigma, band_names) for an arbitrary frequency list."""
    freqs_use = np.asarray(freqs_use, dtype=float)
    model = compute_Dl_model(
        r=params['r'], A_d=params['A_d'], beta_d=params['beta_d'],
        alpha_d=alpha_d_fixed, freqs_use=freqs_use,
    )
    sigma = make_sigma_Dl(model, freqs_use=freqs_use)

    rng_local = np.random.default_rng(seed)
    z = rng_local.standard_normal(model.size)
    n_dof_t = model.size - 3
    z *= np.sqrt(n_dof_t / np.sum(z**2))
    data = model + z.reshape(model.shape) * sigma
    return data, sigma, freqs_use


def log_posterior_for(theta, data, sigma, freqs_use):
    r, A_d, beta_d = theta
    if not (0.0 <= r <= 0.06 and 0.0 <= A_d <= 80.0 and 1.0 <= beta_d <= 2.2):
        return -np.inf
    model = compute_Dl_model(r=r, A_d=A_d, beta_d=beta_d,
                             alpha_d=alpha_d_fixed, freqs_use=freqs_use)
    chi2 = np.sum(((data - model) / sigma)**2)
    return -0.5 * chi2


configs = {
    '2 bands (no dust mon.)': np.array([93., 145.]),
    '3 bands (baseline)':     np.array([93., 145., 220.]),
    '4 bands (+ low freq)':   np.array([50., 93., 145., 220.]),
}

n_steps_ex6 = 6000
burn_ex6 = 1500
step_ex6 = 0.4 * np.array([0.005, 0.6, 0.02])

fig, ax = plt.subplots(figsize=(7, 4.5))
ex6_summary = {}

for k, (label, freqs_use) in enumerate(configs.items()):
    data_k, sigma_k, freqs_k = make_toy_for_freqs(freqs_use)
    n_spec = len(freqs_k) * (len(freqs_k) + 1) // 2

    def logp(theta, _d=data_k, _s=sigma_k, _f=freqs_k):
        return log_posterior_for(theta, _d, _s, _f)

    chain_k, acc_k = run_mcmc(
        logp,
        start=np.array([params_true['r'], params_true['A_d'], params_true['beta_d']]),
        step_size=step_ex6,
        n_steps=n_steps_ex6,
        seed=20 + k,
    )
    samp = chain_k[burn_ex6:, 0]  # r only
    med, lo, hi = np.median(samp), *np.percentile(samp, [16, 84])
    ex6_summary[label] = (med, lo, hi, acc_k, n_spec, len(freqs_k))

    ax.hist(samp, bins=35, density=True, alpha=0.45, label=fr'{label}: $r={med:.4f}^{{+{hi-med:.4f}}}_{{-{med-lo:.4f}}}$')

ax.axvline(params_true['r'], color='k', ls='--', lw=2, label='true r')
ax.set_xlabel(r'$r$')
ax.set_ylabel(r'$p(r\mid d)$')
ax.set_title(r'Exercise 6: $r$ posterior vs frequency coverage')
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"{'config':28s}  n_bands  n_spectra  acc   r median [16%, 84%]")
for label, (med, lo, hi, acc, n_spec, n_b) in ex6_summary.items():
    print(f'{label:28s}  {n_b:7d}  {n_spec:9d}  {acc:4.2f}  '
          f'{med:.5f} [{lo:.5f}, {hi:.5f}]')

print(
    '\nDiscussion:\n'
    '  - Removing the high-frequency dust-monitoring band broadens p(r|d):\n'
    '    dust amplitude/index are less constrained, so more of the BB power\n'
    '    can be traded into/out of r (stronger r–dust degeneracy).\n'
    '  - A noisy high-frequency channel still helps: it has a large dust SED\n'
    '    lever arm, so even large noise leaves useful information on A_d, beta_d.\n'
    '  - Adding bands increases the number of unique spectra n_freq(n_freq+1)/2\n'
    '    and usually narrows the r posterior if the new band is informative.\n'
    '  - Injecting r=0 yields an upper limit (posterior against the r=0 boundary);\n'
    '    larger injected r yields a detection (peak away from 0).\n'
    '  - Lower A_lens (delensing) reduces the CMB BB floor and typically\n'
    '    tightens the r constraint once foregrounds are controlled.'
)
````
